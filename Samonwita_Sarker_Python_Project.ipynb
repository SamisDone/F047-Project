```python
import pandas as pd
```

```python
data = pd.read_excel("/content/basic_python_mcq_questions.xlsx")
```

```python
data
```

```python
score = 0
total_questions = len(data)
```

```python
for index, row in data.iterrows():
    print("-------------------------------")
    print("Question No:", index + 1)
    print(row["question_text"])
    print("A:", row["option_a"])
    print("B:", row["option_b"])
    print("C:", row["option_c"])
    print("D:", row["option_d"])
    print("-------------------------------")

    user_answer = input("Choose correct option (A/B/C/D): ").strip().upper()

    if user_answer == row["correct_answer"]:
        print("✅ Correct Answer")
        score += 1
    else:
        print("❌ Wrong Answer")
    print("Explanation:", row["answer_explanation"])
    print()
```

```python
print("Quiz Completed!")
print("Your Score:", score, "/", total_questions)
```
